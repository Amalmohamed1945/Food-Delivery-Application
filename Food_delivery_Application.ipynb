```python
# STEP 1: Install required libraries
!pip install plotly pandas --quiet

# STEP 2: Import libraries
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# STEP 3: Load the dataset
file_path = '/content/food_delivery_cleaned_dataset.csv'  # Adjust path if needed
df = pd.read_csv(file_path)

# STEP 4: Quick look at the data
print(df.shape)
print(df.columns)
print(df.dtypes)
df.head()

# STEP 5: Handling missing values (if any)
df = df.dropna()

# STEP 6: Distribution plots for numeric columns
numeric_cols = df.select_dtypes(include='number').columns.tolist()
for col in numeric_cols:
    fig = px.histogram(df, x=col, nbins=20, title=f'Distribution of {col}')
    fig.show()

# STEP 7: Count plots for categorical columns
categorical_cols = df.select_dtypes(include='object').columns.tolist()
for col in categorical_cols:
    fig = px.histogram(df, x=col, title=f'Frequency of {col}', color=col)
    fig.update_layout(xaxis={'categoryorder':'total descending'})
    fig.show()

# STEP 8: Sunburst plot for multilevel category summary
if {'City', 'Type_of_order', 'Festival'}.issubset(df.columns):
    fig = px.sunburst(df, path=['City', 'Type_of_order', 'Festival'], title='City → Type of Order → Festival')
    fig.show()

# STEP 9: 3D Scatter Plot (if applicable)
if {'Delivery_person_Age', 'Delivery_person_Ratings', 'Time_taken(min)'}.issubset(df.columns):
    fig = px.scatter_3d(
        df,
        x='Delivery_person_Age',
        y='Delivery_person_Ratings',
        z='Time_taken(min)',
        color='City',
        title='3D Scatter: Age vs Ratings vs Time Taken'
    )
    fig.show()
```
